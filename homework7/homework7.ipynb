```python
# homework7.ipynb

import numpy as np
import matplotlib.pyplot as plt

# 2: Matplotlib

def make_sine_wave(x, A, w):
    y = A * np.sin(w * x)
    return y

x = np.linspace(0, 2 * np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, axs = plt.subplots(1, 3, figsize=(25, 5))

for i in range(len(amplitude)):
    A = amplitude[i]
    w = 1
    y = make_sine_wave(x, A, w)
    axs[0].plot(x, y, label="A =" + str(A))

axs[0].set_title("Varying Amplitude (w = 1)", fontsize=18)
axs[0].set_xlabel("x", fontsize=14)
axs[0].set_ylabel("y = A·sin(w·x)", fontsize=14)
axs[0].legend(fontsize=12)

for j in range(len(frequency)):
    A = 1
    w = frequency[j]
    y = make_sine_wave(x, A, w)

    axs[1].plot(x, y, label="w = " + str(w))

axs[1].set_title("Varying Frequency (A = 1)", fontsize=18)
axs[1].set_xlabel("x", fontsize=14)
axs[1].set_ylabel("y = A·sin(w·x)", fontsize=14)
axs[1].legend(fontsize=12)

for k in range(len(amplitude)):
    A = amplitude[k]
    w = frequency[k]
    y = make_sine_wave(x, A, w)

    axs[2].plot(x, y, label="A = " + str(A) + ", w = " + str(w))

axs[2].set_title("Varying Amplitude and Frequency", fontsize=18)
axs[2].set_xlabel("x", fontsize=14)
axs[2].set_ylabel("y = A·sin(w·x)", fontsize=14)
axs[2].legend(fontsize=12)

plt.show()

# 3: Plot a Hertzsprung-Russell Diagram!

```
