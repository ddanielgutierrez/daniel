```python
import pandas as pd

df = pd.read_csv('../../course_assignments/homework8/state_data.csv')

# 2.4: 52 rows, 4 columns
df.shape
print("Shape:", df.shape)

# 2.5: 1st Column: State   2nd Column: Latitude  3rd Column: Longitude   4th Column: State full name
# Alaska
# Wyoming
# 51
# Includes U.S Territory?

df.head()
print(df.head())

df.tail()
print(df.tail())

median_latitude = df['latitude'].median()
median_longitude = df['longitude'].median()

print("The median latitude is: ", median_latitude)
print("The median longitude is: ", median_longitude)

df['latitude_distance'] = abs(df['latitude'] - median_latitude)
df['longitude_distance'] = abs(df['longitude'] - median_longitude)

print(df['latitude_distance'])
print(df['longitude_distance'])

print(df.shape)
# 2.9: 52 rows, 6 columns

# 2.10
closest_latitude = df.loc[df['latitude_distance'].idxmin()]
farthest_latitude = df.loc[df['latitude_distance'].idxmax()]

print("Closest latitude is: ", closest_latitude)
print("Farthest latitude is: ", farthest_latitude)

closest_longitude = df.loc[df['longitude_distance'].idxmin()]
farthest_longitude = df.loc[df['longitude_distance'].idxmax()]

print(df.sample(2))
# Home -> API reference -> Dataframe -> pandas.DataFrame.Sample
# Return a random sample of items from an axis of object. Useful for large datasets

# 3.0: Seaborn

import matplotlib.pyplot as plt
import seaborn as sns

df2 = sns.load_dataset('flights')

# 3.5:

df2 = df2.dropna(axis=1)

# 3.6 a) 144 rows, 3 columns
print(df2.shape) # (144, 3)

print(df2.head())
print(df2.tail())
print(df2.columns)
# d) 
# Year: Year of Record
# Month: Month of Record
# Passengers: Number of Passengers onboard

# 3.7: Create a Subplot

fig, axs = plt.subplots(2, 2, figsize = (12, 10))

sns.scatterplot(data=df2, x = 'year', y = 'passengers', ax = axs[0, 0], color = 'red')
axs[0, 0].set_title('Passengers vs Year')
axs[0, 0].set_xlabel('Year')
axs[0, 0].set_ylabel('Passengers')

df2['month_numb'] = df2['month'].astype('category').cat.codes

print(sns.scatterplot)

sns.scatterplot(data = df2, x = 'month', y = 'passengers', ax = axs[0, 1], color = 'green')
axs[0, 1].set_title('Passengers vs Month')
axs[0, 1].set_xlabel('Month')
axs[0, 1].set_ylabel('Passengers')

sns.histplot(data = df2, x = 'passengers', bins = 10, ax = axs[1,0], color = 'blue')
axs[1, 0].set_title('Passenger Histogram')
axs[1, 0].set_xlabel('Passengers')
axs[1, 0].set_ylabel('Count')

sns.boxplot(data = df2, x = 'year', y = 'passengers', ax = axs[1, 1], color = 'yellow')
axs[1, 1].set_title('Passenger Count per Year')
axs[1, 1].set_xlabel('Year')
axs[1, 1].set_ylabel('Count')

plt.tight_layout()
plt.show()

#3.9: Explain
# 1: Passenger count per year. Shows the increase in passengers over the course of 10 years, which shows that there has been a steady increase since the 1950's. Can be explained by the advancement in technology
# 2: Passenger count per month. Can confidently say that during the month of July and August, there are the most amount of passengers. A potential explanation for this is summer vacations.
# 3: Shows how passenger counts are distributed. Most months fall in the 200–350 range, with fewer very low or very high values.
# 4: Shows growth year-by-year, with medians increasing over time and some years having a wider spread in monthly passenger counts.
```
