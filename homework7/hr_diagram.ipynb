```python
# hr_diagram.ipynb

import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
                  skiprows=1)

# 3.2: Analyzing Important Data

print("Data shape:", data.shape)
magnitude = data[:, 0]
temperature = data[:, 1]

print("All magnitudes:", magnitude)
min_magnitude = np.min(magnitude)
max_magnitude = np.max(magnitude)

print("Minimum magnitude:", min_magnitude)
print("Maximum magnitude:", max_magnitude)

index_min_mag = np.argmin(magnitude)
index_max_mag = np.argmax(magnitude)

print("Minimum magnitude is at row:", index_min_mag)
print("Maximum magnitude is at row:", index_max_mag)

min_temperature = np.min(temperature)
max_temperature = np.max(temperature)

index_min_temp = np.argmin(temperature)
index_max_temp = np.argmax(temperature)

print("Minimum temperature:", min_temperature, "at row", index_min_temp)
print("Maximum temperature:", max_temperature, "at row", index_max_temp)

brightest_index = index_min_mag
dimmest_index = index_max_mag

print("Brightest star index:", brightest_index)
print("Dimmest star index:", dimmest_index)

hottest_index = index_max_temp
coldest_index = index_min_temp

print("Hottest star index:", hottest_index)
print("Coldest star index:", coldest_index)

# Brightest star = smallest magnitude
# Dimmest star = largest magnitude
# Hottest star = largest temperature
# Coldest star = smallest temperature

# 3.3: Plot and HR Diagram

log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))

plt.scatter(log_temperature, log_magnitude,
            color="blue", s=20, label="Stars")

plt.scatter(log_temperature[brightest_index],
            log_magnitude[brightest_index],
            color="black", marker="*", s=300,
            label="Brightest")

plt.scatter(log_temperature[dimmest_index],
            log_magnitude[dimmest_index],
            color="purple", marker="*", s=300,
            label="Dimmest")

plt.scatter(log_temperature[hottest_index],
            log_magnitude[hottest_index],
            color="green", marker="*", s=300,
            label="Hottest")

plt.scatter(log_temperature[coldest_index],
            log_magnitude[coldest_index],
            color="orange", marker="*", s=300,
            label="Coldest")

ax = plt.gca()
ax.invert_xaxis()
ax.invert_yaxis()

plt.title("Hertzsprung–Russell Diagram", fontsize=16)
plt.xlabel("Temperature (log K)", fontsize=14)
plt.ylabel("Magnitude (log scale)", fontsize=14)
plt.legend()

plt.show()

# Bonus

plt.figure(figsize=(8, 6))

scatter_plot = plt.scatter(log_temperature, log_magnitude,
                           c=temperature, s=20)

ax = plt.gca()
ax.invert_xaxis()
ax.invert_yaxis()

plt.title("Hertzsprung–Russell Diagram (Colored by Temperature)", fontsize=16)
plt.xlabel("Temperature (log K)", fontsize=14)
plt.ylabel("Magnitude (log scale)", fontsize=14)

cbar = plt.colorbar(scatter_plot)
cbar.set_label("Temperature (K)")

plt.show()
```
